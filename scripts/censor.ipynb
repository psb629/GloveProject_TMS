```python
import numpy as np
import pandas as pd
from glob import glob
from os.path import join, exists
```

---

```python
dir_work = join('../censor')
```

```python
list_fname = glob(join(dir_work, '*.1D'))
```

```python
res = {'subj':[], 'total':[], 'zero':[], 'ratio':[]}
for fname in sorted(list_fname):
    subj = fname.split('/')[2].split('_')[1]
    with open(fname, "r") as f:
        lines = f.readlines()
    total_count = 0
    zero_count = 0

    for line in lines:
        line = line.strip()
        total_count += len(line)
        zero_count += line.count("0")
    ratio = zero_count / total_count
    res['subj'].append(subj)
    res['total'].append(total_count)
    res['zero'].append(zero_count)
    res['ratio'].append(ratio)
df = pd.DataFrame(res)
df
```

```python
print(df.ratio.mean(), df.ratio.std())
```
